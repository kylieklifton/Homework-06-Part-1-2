## 4a) How many male billionaires are there compared to the number of female billionares?

> **TIP:** This is not a `.groupby`

```python
# Hey dataframe
# How many male billionaires are there compared to the number of female billionaires?
df['gender'].value_counts()
```

## 4b) What percent is that?

> **TIP:** Doing percentages with the answer to `4a` is so common that there's an extra parameter to automatically calculate it. Look at the documentation!

```python
# Hey dataframe
# What percent is that?
df['gender'].value_counts(normalize=True)
# 0.89 (89%) of the billionaires are male.
# 0.10 (10%) of the billionaires are female.
```

## 4c) Is the average wealth different for a female billionaire as compared to a male billionaire?

> **TIP:** This one *does* use `.groupby`. You can pick what 'average' means in this context.

```python
# Hey variable "females"
# Make females an isolated variable for future use representing the 180 female billionaires
females = df[df['gender'] == 'female']
```

```python
# Hey variable "males"
# Make males an isolated variable for future use representing the 1473 male billionaires
males = df[df['gender'] == 'male']
```

```python
# Hey dataframe
# Use groupby
df.groupby('gender')['networthusbillion'].mean()
# The average wealth for female billionaires is 3.92 billion and for 3.90 billion for male billionaires
# However, the number of female billionaires is 0.1221 (12.21%) of male billionaires so this might not be a normalized comparison.
```

## 5a) What is the most common source/type of wealth?

```python
# Hey dataframe
# Determine the most common source/type of wealth
df['typeofwealth'].value_counts()
# The most common source of wealth amongst American billionaires is inherited wealth. 
# 506 American billionaires inherited their wealth, and 467 of them were founders non-finance.
```

```python
# Hey dataframe
# What is the share of most common types of wealth?
df['typeofwealth'].value_counts(normalize=True)
```

## 5b) Is the most common source/type of wealth different between men and women?

> **TIP:** You know how to `groupby` and you know how to count how many times a value is in a column. Can you put them together??? You could also do this as two separate cells where you calculate it separately for men and women.
>
> **TIP:** Use percentages for this, it makes it a lot more readable.

```python
# Hey dataframe
# What is the most common source of wealth for the top 5 wealthiest female billionaires?
females.groupby('gender')['typeofwealth'].value_counts()
# The most common source of wealth for female billionaires is inherited wealth.
```

```python
# Hey dataframe
# What is the most common source of wealth for the top 5 wealthiest male billionaires?
males.groupby('gender')['typeofwealth'].value_counts()
# The most common source of wealth for male billionaires is wealth earned as a founder non-finance.
# So, the most common source/type of wealth is different between male and female american billionaires.
```

## 6a) What companies have the most billionaires? Let's see the top five.

> **TIP:** You can use `.head()` on *anything*, not just your basic `df`

> **TIP:** You might feel like you should use `groupby`, but don't! There's an easier way to count.

## 7) How much money do these top 5 billionaires have in total?

## 8) What are the top 10 countries with the most money held by billionaires?

I am **not** asking which country has the most billionaires - this is **total amount of money per country.**

> **TIP:** Think about it in steps - "I want them organized by country," "I want their net worth," "I want to add it all up," and "I want 10 of them." Just chain it all together.

```python
# Hey dataframe
# Find the top 10 countries with the most money in the hands of billionaires
df.sort_values('networthusbillion', ascending=False)\
  .groupby('citizenship')\
  .head(10)[['citizenship', 'networthusbillion']]

```

## 9a) How old is an average billionaire?

```python
# Hey dataframe
# How old is the average billionaire?
# Use .mean()
df['age'].mean()
# The age of an average billionaire is 63 years old.
```

## 9b) How old are self-made billionaires compared to non self-made billionaires? 

```python
# Hey dataframe
# How old the are self-made billionaires compared to the non self-made?
df.groupby('selfmade')['age'].mean()
# The average age of self-made billionaires is 2 years younger than non self-made billionaires.
# The average age of inherited billionaires is 65 years old.
# The average age of self-made billionaires is 63 years old.
```

```python
# Hey dataframe
# Compare the median, as average might be deterred by outliers
df.groupby('selfmade')['age'].median()
# The median age of self-made billionaires is 2 years younger than non self-made billionaires.
# The median age of inherited billionaires is 65 years old.
# The median age of self-made billionaires is 63 years old.
```
